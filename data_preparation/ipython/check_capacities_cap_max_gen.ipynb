```python
import os
import pandas as pd
import datetime as dt
```

```python
dir_out = "../parsed_data/"
fn_gen_in = "generation_2017_hourly_entsoe.csv"
fn_cap_in = "capacities.csv"
```

```python
df_gen_hourly = pd.read_csv(dir_out + fn_gen_in)
df_gen_hourly_max = df_gen_hourly.groupby(['country','tech']).max().drop(columns={'date'})
df_gen_hourly_max.head()
```

```python
df_cap = pd.read_csv(dir_out + fn_cap_in)
df_cap = df_cap[df_cap.year==2017].drop(columns={'year'}).rename(columns={'technology':'tech'}).set_index(['country','tech'])
df_cap.head()
```

```python
df_compare = df_gen_hourly_max.join(df_cap)
```

```python
df_compare['share'] = df_compare['output']/df_compare['capacity']
df_compare['scale'] = df_compare['capacity']/df_compare['output']
```
