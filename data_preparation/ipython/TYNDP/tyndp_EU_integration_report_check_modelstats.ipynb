## TYNDP EU integration report to check if all scenarios solved

```python
import pandas as pd
import numpy as np
import gdxtools as gt
import plotly.express as px
import plotly
import gams
import os
import sys
```

```python
path = "C:/Users/jonas/Documents/eu_electricity_model/result_dashboard/gdx/TYNDP/"
files = os.listdir(path)
files = [ele for ele in files if '.gdx' in ele]
```

```python
dir_gms = os.getcwd()
ws = gams.GamsWorkspace(dir_gms)
```

# load model stats

```python
#load gdx data on load and assign periods for ntc on files
df_stats = pd.DataFrame()
for file in files:
    gdx = ws.add_database_from_gdx(path+file)
    df_temp = pd.DataFrame(gt.get_symbol_values(gdx, "r_modelstatistics")).reset_index()
    df_temp['scenario'] = file[7:-5]
    df_stats = df_stats.append(df_temp)
```

```python
df_stats = df_stats.pivot_table(index='scenario',columns='dim1',values='Value')
```
