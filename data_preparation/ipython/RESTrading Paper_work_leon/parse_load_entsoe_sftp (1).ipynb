```python
#Leon
import chardet
# Eine der Datein testen
test_file = path_load_local + files[0]
print("Prüfe Datei:", test_file)

# Encoding automatisch erkennen
with open(test_file, 'rb') as f:
    result = chardet.detect(f.read(50000)) # nur ersten Teil lesen
print("Erkanntes Encoding:", result)

# Datei mit erkanntem Encoding laden
df_test = pd.read_csv(test_file, sep="\t", encoding=result['encoding'], nrows=5)
print("Spalten:", df_test.columns.tolist())

```

```python
#combine files to one data frame
df_load = pd.DataFrame()
for file in files:
    df_temp = pd.read_csv(path_load_local+file,
                          decimal=".",encoding="UTF-8-SIG",sep="\t",  # Leon: updated encoding
                          parse_dates=True, index_col="DateTime").drop("AreaCode", axis=1)
    #df_load = df_load.append(df_temp)
    df_load = pd.concat([df_load, df_temp], ignore_index=False) # Leon, in pandas 2.0 gibt es .append() nicht mehr als Methode
df_load = df_load[df_load.AreaTypeCode == "CTY"].drop(["AreaTypeCode",'ResolutionCode','AreaName','UpdateTime'], axis=1).reset_index()
df_load = df_load.sort_values(by=['DateTime'])
df_load.info()
```

```python
df_load = df_load.rename(columns={"MapCode": "country", 'DateTime':'date','TotalLoadValue':'load'})
df_load = df_load.pivot_table("load", "date", "country")
```
