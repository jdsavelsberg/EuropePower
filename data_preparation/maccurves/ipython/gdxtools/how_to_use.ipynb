# Packages and options

```python
%load_ext autoreload
%autoreload 2
import numpy as np
import itertools
import os
from GdxFile import GdxFile
```

```python
fn = "./test.gdx"
```

# Create empty gdx and add sets

Create empty gdx file

```python
gdx = GdxFile(fn)
```

Add some sets

```python
i = ["i%d" % i for i in range(1,101)]
r = ["r%d" % i for i in range(1,11)]
gdx.add_set(i, "i")
gdx.add_set(r, "r")
```

Add a scalar

```python
gdx.add_parameter(1, "test_scalar")
```

Add one and two dimensional parameter

```python
gdx.add_parameter({j: np.random.uniform(0,100) for j in i}, "one_dim")
```

```python
gdx.add_parameter({j: np.random.uniform(0,100) for j in itertools.product(i,r)}, "two_dim")
```

```python
gdx.save()
```

```python
del gdx
```

# Reading a gdx file

```python
gdx = GdxFile(fn)
```

We have two possibilities to read values. 
1. Using [] assessors we get a series (list) in case of parameters, variables, etc (set)
   names of indexes are inferred from gdx file. I.e. if domain is explicitly set, we get the GAMS name of the dimension as index name
2. Using the get_symbol method, we can name the dimensions of the pandas series

Reading the sets

```python
i,r = gdx["i"], gdx["r"]
r
```

Scalars are returned as scalar values:

```python
gdx["test_scalar"]
```

All parameters are returned as pandas series possibly with mulit-index for the respective set dimension

```python
gdx["one_dim"].head()
```

```python
gdx["two_dim"]
```

By default, dimensions names are inferred from the GAMS names of the set symbols. In the example above it was the universal set '\*' as we did use domain checking. Using *get_symbol*, we can set dimension names explicitly: 

```python
gdx.get_symbol("two_dim", col_names=["item", "region"])
```

# Delete created gdx file

First free the GdxFile (and with that the underlying GamsDatabase) then use standard Python deletion.

```python
del gdx
os.remove(fn)
```
